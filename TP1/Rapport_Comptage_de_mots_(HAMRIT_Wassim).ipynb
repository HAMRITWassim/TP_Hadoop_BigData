# **Rapport - Utilisation de plusieurs méthodes pour compter l'occurence des mots d'un texte**


Dans ce rapport, on va étudier plusieurs méthodes permettant de compter les mots dans un fichier texte. Le problème en Big Data est que le traitement de très grandes données (plusieurs GigaOctets voire TeraOctets) peut être très long selon la méthode utilisée. <br><br>
On va s'intéresser à **3 méthodes** différentes: 

- Une méthode simple en utilisant du code Python en traitant les données de manière **séquetielle**.
- Une méthode **pseudo-MapReduce** qui reproduit les étapes du MapReduce (map, shuffle, reduce) en utilisant Python.
- Une méthode qui s'appuie sur le parallélisme en utilisant **Hadoop MapReduce**.

Pour chacune de ces méthodes, on va utiliser plusieurs fichiers texte en faisant varier leur nombre de lignes:
- **N = 100 lignes** 
- **N = 100 000 lignes** 
- **N = 5 000 000 lignes** 
- **N = 50 000 000 lignes** 



En utilisant ces fichiers texte de différentes tailles, on pourra donc étudier les différences de performance entre chaque méthodes selon la quantité des données.

## **1 - Méthode séquentielle avec Python**


#### a) Fichier de taille N = 100 lignes

````python
import time

def compter_occurrences(fichier):
    with open (fichier , 'r', encoding='utf -8') as f:
        mots = f.read().split()

    compteur = {}
    for mot in mots:
        compteur[mot] = compteur.get(mot , 0) + 1
    return compteur

debut = time.perf_counter();

fichier = "./textes/fichier1_100.txt"
resultats = compter_occurrences(fichier)
for mot , nb in sorted (resultats.items(), key= lambda x: x[1], reverse=True):
    print (f"{ mot} : {nb}")

fin = time.perf_counter();
    
temps_exec = fin - debut

print(f"\nTemps d'exécution : {temps_exec:.5f}s")
````

## **2 - Méthode pseudo-MapReduce avec Python**


#### a) Fichier de taille N = 100 lignes

````python
import time
from collections import defaultdict

def mapper(ligne):
    mots = ligne.strip().split()
    return [(mot, 1) for mot in mots]

def shuffle(mapped_data):
    grouped = defaultdict(list)
    for mot, valeur in mapped_data:
        grouped[mot].append(valeur)
    return grouped

def reducer(grouped_data):
    reduced = {}
    for mot, valeurs in grouped_data.items():
        reduced[mot] = sum(valeurs)
    return reduced

debut = time.perf_counter();

fichier = "./textes/fichier1_100.txt"
mapped = []
    
with open(fichier, "r", encoding="utf-8") as f:
    for ligne in f:
        mapped.extend(mapper(ligne))

grouped = shuffle(mapped)
reduced = reducer(grouped)

for mot, count in sorted(reduced.items(), key=lambda x: x[1], reverse=True):
    print(f"{mot} : {count}")


fin = time.perf_counter();
    
temps_exec = fin - debut

print(f"\nTemps d'exécution : {temps_exec:.5f}s")
````

## **5 - Méthode PySpark** 

La principale différence entre PySpark et Hadoop MapReduce est le traitement de la mémoire :
Tandis que Hadoop MapReduce va écrire les résultats intermédiaires sur le disque dur (entre les étapes Map et Reduce), ce qui peut ralentir le processus, PySpark va garder les données en RAM et les découper dans des structures appelées RDD (Resilient Distributed Datasets), permettant de réduire les accès au disque, et donc d'obtenir de meilleures performances en exploitant la parallélisation.

#### a) Fichier de taille N = 100 lignes

````python
import sys, os, time
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable
os.environ["JAVA_HOME"] = r"C:\Program Files\Microsoft\jdk-17.0.20.101-hotspot"
os.environ["HADOOP_HOME"] = r"C:\hadoop"

from pyspark.sql import SparkSession, functions as F

spark = SparkSession.builder.master("local[*]").appName("WordCount").getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

def wordcount_df(chemin):
    return (spark.read.text(chemin)                                         # 1 ligne = 1 valeur
                 .select(F.explode(F.split("value", r"\s+")).alias("mot"))  # MAP : lignes -> mots
                 .where(F.col("mot") != "")
                 .groupBy("mot").count()                                     # SHUFFLE + REDUCE
                 .collect())

wordcount_df("./textes/fichier1_100.txt")   # échauffement


debut = time.perf_counter()
resultats = wordcount_df("./textes/fichier1_100.txt") 
temps_exec = time.perf_counter() - debut
print(f"Temps d'exécution PySpark (DataFrame) : {temps_exec:.2f} s")
print(sorted([(r["mot"], r["count"]) for r in resultats], key=lambda x: x[1], reverse=True))



spark.stop()
````

## **6 - Comparaison des 4 méthodes :**

| Nombre de lignes | Fichier | Séquentiel Python | Pseudo-MapReduce Python | Hadoop MapReduce | PySpark |
| :--- | :--- | :---: | :---: | :---: | :---: |
| **100** | `fichier1_100.txt` | 0.01228 s | 0.00083 s | 4.766 s | 0.15 s |
| **100 000** | `fichier2_100k.txt` | 0.04713 s | 0.06905 s | 3.236 s | 0.19 s |
| **5 000 000** | `fichier3_5M.txt` | 0.71780 s | 2.48146 s | 5.229 s | 0.71 s |
| **50 000 000** | `fichier4_50M.txt` | 7.60132 s | 23.38727 s | 25.537 s | 8.71 s |

En passant par un DataFrame et en intégrant une phase de chauffe, PySpark réussit a avoir de meilleurs résultats qu'Hadoop MapReduce en éliminant notamment les accès disque en utilisant la mémoire pour traiter les données. Sur ces données, Python reste au niveau car les méthodes comme Hadoop MapReduce & PySpark subissent des ralentissements à cause des démarrages. Ces méthodes pourraient être utiles pour des données beaucoup plus importantes (plusieurs GigaOctets), où la méthode Python peut subir des erreurs de mémoire.

````python
import matplotlib.pyplot as plt

# Données issues du tableau
lignes = [100, 100000, 5000000, 50000000]
seq_python = [0.01228, 0.04713, 0.71780, 7.60132]
pseudo_mr = [0.00083, 0.06905, 2.48146, 23.38727]
hadoop_mr = [4.766, 3.236, 5.229, 25.537]
pyspark = [0.15, 0.19, 0.71, 8.71]

# Création du graphique
plt.figure(figsize=(10, 6))

# Tracé des différentes méthodes
plt.plot(lignes, seq_python, marker='o', linewidth=2.5, label='Séquentiel Python')
plt.plot(lignes, pseudo_mr, marker='s', linewidth=2.5, label='Pseudo-MapReduce Python')
plt.plot(lignes, hadoop_mr, marker='^', linewidth=2.5, label='Hadoop MapReduce')
plt.plot(lignes, pyspark, marker='D', linewidth=2.5, label='PySpark (DataFrame)')

# Configuration des axes 
plt.xscale('log')
plt.xlabel("Nombre de lignes", fontweight='bold')
plt.ylabel("Temps d'exécution (secondes)", fontweight='bold')
plt.title("Comparaison des performances des méthodes de WordCount", fontweight='bold')


plt.legend(loc="upper left")
plt.grid(True, which="both", linestyle="--", alpha=0.5)


plt.show()
````
